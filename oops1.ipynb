```python
# Create a simple banking app
class BankingApp:
    def __init__(self):
        self.pin = ""
        self.balance = 0
        self.menu()
    
    def menu(self):
        user_input = int(input("""
              Welcome to the atm
              1. Create pin
              2. Change pin
              3. Check Balance
              4. Withdraw money
              5. Exit"""))
        
        if user_input == 1:
            self.create_pin()
        elif user_input == 2:
            self.change_pin()
        elif user_input == 3: 
            self.display_balance()
        elif user_input == 4: 
            self.withdraw()
        elif user_input == 5:
            exit()
        else:
            print("Invalid Input")
            self.menu()
            
    def create_pin(self):
        self.pin = int(input("Enter a pin: "))
        self.balance = int(input("Enter your balance: "))
        print("Pin set!")
        self.menu()
    
    def change_pin(self):
        verify_pin = int(input("Enter your current pin: "))
        if(self.pin == verify_pin):
            self.pin = int(input("Enter your new pin: "))
            print("Pin reset!")
            self.menu()
        else:
            print("Wrong Pin!")
            self.menu()
    
    def display_balance(self):
        print(f"Balance: {self.balance}")
        self.menu()
    
    def withdraw(self):
        input_pin = int(input("Enter your pin: "))
        if input_pin == self.pin:
            amt = int(input("Enter amount to withdraw: "))
            if amt < self.balance:
                self.balance -= amt
                print(f"{amt} is withdrawn")
                self.menu()
            else:
                print("You do not have enough balance!")
                self.menu()
        else:
            print("Invalid pin")
            self.menu()
```

Magic (Dunder) methods

__methodname__ -> like __init__

magic methods dont need to be called explicitly

Use of constructor: Put configuration related code in constructor as the constructor is not under the control of the user as opposed to other methods

```python
class Fraction:
    
    def __init__(self, x, y):
        self.numerator = x
        self.denominator = y
        
    def __str__(self):
        return(f"{self.numerator}/{self.denominator}")
    
    # add is called when you do obj1 + obj2 
    def __add__(self, other):
        return (f"{self.numerator*other.denominator + other.numerator*self.denominator}/{self.denominator*other.denominator}")
    
    def __sub__(self, other):
        return (f"{self.numerator*other.denominator - other.numerator*self.denominator}/{self.denominator*other.denominator}")
    
    def __mul__(self,other):
        return (f"{self.numerator*other.numerator}/{self.denominator*other.denominator}")
    
fr1 = Fraction(1,2)
fr2 = Fraction(3,4)
print(fr1)
print(fr1 + fr2)
print(fr2 - fr1)
```

```python
class Coordinate:
    def __init__(self, x, y):
        self.coord = (x,y)
    
    def __str__(self):
        return str(self.coord)
    
    def dist(self, other):
        return ((self.coord[0] - other.coord[0])**2 + (self.coord[1] - other.coord[1])**2)**0.5
    
    def dist_from_origin(self):
        return ((self.coord[0])**2 + (self.coord[1])**2)**0.5
    
    
    
```

```python
p1 = Coordinate(1,1)
p2 = Coordinate(1,2)
print(p1)
print(p2)
print(p1.dist(p2))
```

```python
class Line:
    def __init__(self, x_coeff, y_coeff, const):
        self.coeffs = (x_coeff, y_coeff, const)
        
    def __str__(self):
        return (f"{self.coeffs[0]}x + {self.coeffs[1]}y + {self.coeffs[2]} = 0")

    def point_on_line(self, point):
        if self.coeffs[0]*point.coord[0] + self.coeffs[1]*point.coord[1] + self.coeffs[2] == 0:
            return True
        else:
            return False
        
    def closest_dist(self, point):
        return abs(self.coeffs[0]*point.coord[0] + self.coeffs[1]*point.coord[1] + self.coeffs[2]) / (self.coeffs[0]**2 + self.coeffs[1]**2)**0.5
        
        
        
        
```

```python
l1 = Line(2,3, 5)
p3 = Coordinate(-1,-1)
print(l1)
print(l1.point_on_line(p3))
print(l1.closest_dist(p3))
```
