```python
# Serializing a list
import json
L = [1,2,3,4]

with open('demo.json', 'w') as f:
    json.dump(L, f)
```

```python
d = {"name": "Vihaan",
     "age": 19,
     "gender": "Male"}

with open('demo.json', 'w') as f:
    json.dump(d, f, indent=4)
```

```python
with open('demo.json', 'r') as f:
    d = json.load(f)
    print(d)
```

```python
# serializing your own object
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    
# show person can return a dict as well - it doesnt have to be a string output
def show_person(person):
    if isinstance(person, Person):
        return f"Name: {person.name} Age: {person.age} Gender: {person.gender}"

p = Person("Vihaan", 19, "Male")
with open("demo.json", 'w') as f:
    json.dump(p, f, default = show_person)
    
with open('demo.json', 'r') as f:
    d = json.load(f)
    print(d)
```
